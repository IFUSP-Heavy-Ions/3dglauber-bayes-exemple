## Modelo de correlacao

CS: rho = 1; correlacao uniforme dentro de cada setor.

Cada chave exata de observavel e sistema constitui um bloco. Nos observaveis eta configurados, o centro e um setor e as duas laterais, juntas, sao outro. Nao ha covariancia experimental entre esses setores, nem entre observaveis ou sistemas distintos. No RBF, as laterais compartilham o setor, mas a correlacao depende da distancia fisica em eta. A cobertura continua PHOBOS dN/deta permanece em um unico bloco.

A covariancia do emulador continua incluida no ajuste. CS com rho=1 gera blocos experimentais semidefinidos, nao um erro por si so.

Detalhes e tamanhos de cada setor na reconstrucao local: `covariance_check.json`. Os metadados salvos na cadeia sao a referencia para identificar o ajuste original.

## Validacao e limites

As medias dos emuladores carregados foram comparadas com os quantis salvos do estudo global CS: **40/40 observaveis** coincidiram (rtol=1e-6, atol=1e-10). A classe atual de `emulator.py` foi ligada ao estado treinado, sem retreinar, evitando depender de bytecode Python legado do cache dill. As versoes usadas e os hashes dos emuladores constam em `study_manifest.json`.

Os caches foram salvos com scikit-learn 1.6.1 e lidos com 1.8.0. O teste historico valida as medias usadas nestes graficos, nao uma equivalencia geral entre versoes ou a covariancia preditiva.

Assinatura binaria da covariancia local igual a da cadeia: **False**. Se diferente, isso nao permite certificar a reproducao bit a bit da verossimilhanca remota; nao foi determinado se a causa e arredondamento entre ambientes ou alguma diferenca nos inputs. As regioes atuais foram comparadas com as salvas e coincidem. Nenhum novo ajuste, reponderacao ou alteracao da cadeia foi feito. As predicoes usam diretamente os parametros da cadeia salva.

A cadeia foi inspecionada quanto a forma e valores finitos da amostra retida. Nao foram calculados ESS, tempo de autocorrelacao ou R-hat; a finalizacao da cadeia nao demonstra convergencia.
