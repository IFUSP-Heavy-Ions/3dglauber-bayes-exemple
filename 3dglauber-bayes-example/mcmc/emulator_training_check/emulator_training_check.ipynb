# Emulator training-point reproduction

This report collects all 40 pages of the saved training-check PDFs: 34 observables for Au–Au at 200 GeV and 6 for d–Au at 200 GeV. All figures are embedded, so the notebook can be read or published without the PDFs or the original simulation files.

The emulator is evaluated at the same parameter points used for training. This is an in-sample reproduction check (pseudovalidation), not a test on independent simulations and not a Bayesian parameter-recovery closure test.

## How to read the figures

The left panel compares each simulation value with the emulator prediction. Agreement follows the dashed diagonal. Each point represents one training-design point and one observable bin.

The right panel shows prediction minus simulation, sorted by simulation value. Purple bands show the emulator’s central Gaussian predictive intervals: 60% (darker) and 90% (lighter). The reported coverage is the fraction of residuals inside each band, pooled over the design points and bins of that observable. It is not the fraction of entire design vectors covered simultaneously.

For multiplicity observables (`dN`), both predictions and simulation values use the logarithmic scale employed in training. Flow and mean transverse momentum retain their configured observable coordinates. The plotted uncertainty comes from the emulator, rather than an experimental-error band.

## Observed coverage

Across the saved observables, coverage of the nominal 60% band ranges from **67.5% to 89.1%** for Au–Au and **74.4% to 87.1%** for d–Au. Coverage of the nominal 90% band ranges from **93.7% to 99.4%** and **95.7% to 98.7%**, respectively. These are ranges across observables, not a combined coverage estimate.

Every reported coverage is above its nominal level. On training data, this does not establish that uncertainty is calibrated for new points. Use the scatter and residual panels to inspect systematic offsets and individual outliers. Independent validation is needed to assess prediction away from training points.
