## Reference vs. baseline posterior comparison

The corner plot overlays all 20 parameters from the saved reference and baseline production chains. Green shows the baseline; dashed purple contours show the reference. The 2D contours enclose 68% and 90% probability, estimated with the same histogram bins and smoothing. Diagonal panels compare marginal densities; titles summarize the baseline median and central 90% interval.

Both chains are sampled with a step stride of at least 50 and a cap of approximately 200,000 samples each. No additional production burn-in is removed. The chain dimensions and retained sample counts are recorded in `plots/reference_vs_baseline_corner.json`. A common parameter order is assumed from the design-range file, as in the existing plots.

The marginal 90% intervals overlap for **20 of 20 parameters**. The table below allows direct comparison of the medians and interval widths. ## Emulator covariance and correlation maps

These are emulator-error correlations conditional on walker 0 at the final saved step (index 49,999). They are not experimental-error correlations or averages over the posterior. The saved covariance arrays have shapes (1, 791, 791) for Au–Au and (1, 123, 123) for d–Au.

Each map shows covariance normalized by the corresponding standard deviations. The diagonal is one; off-diagonal values describe how emulator errors vary together across bins. Observable names identify the collision system, measurement, centrality, and detector. The combined article figure repeats selected individual maps. Metadata contains the malformed parameter-label list `N, o, n, e`; do not use it to identify the 20 chain parameters.
